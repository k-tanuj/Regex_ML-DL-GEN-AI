```python
import numpy as np
import pandas as pd
```

```python
import tensorflow as tf
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.metrics import accuracy_score
```

```python
df = pd.read_csv('Day 10 Practise 2nd\\covid_toy.csv')
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
df['fever'] = df['fever'].fillna(df['fever'].mean())
```

```python
df.isnull().sum()
```

```python
lb = LabelEncoder()
```

```python
df['gender'] = lb.fit_transform(df['gender'])
df['cough'] = lb.fit_transform(df['cough'])
df['city'] = lb.fit_transform(df['city'])
df['has_covid'] = lb.fit_transform(df['has_covid'])
```

```python
df.head()
```

```python
x = df.drop(columns=['has_covid'], axis=1)
y= df['has_covid']
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
sc= StandardScaler()
```

```python
x_train_scaled =sc.fit_transform(x_train)
x_test_scaled = sc.fit_transform(x_test)
```

```python
model = tf.keras.Sequential([
    tf.keras.layers.Dense(units = 128 ,activation='relu' ,input_dim=x_train_scaled.shape[1]), #input layer
    tf.keras.layers.Dropout(0.3),
    tf.keras.layers.Dense(units=64 ,activation='relu'), #hidden layer
    tf.keras.layers.Dropout(0.3),
    tf.keras.layers.Dense(units = 1 ,activation='sigmoid') #output layer
    ])
```

```python
loss_fn = tf.keras.losses.BinaryCrossentropy()
metrics = ['accuracy']
```

```python
learning_rate = 0.001
momentum = 0.9
optimizer = tf.keras.optimizers.SGD(
    learning_rate=learning_rate, 
    momentum=momentum,
    nesterov=True
    )
```

```python
model.compile(optimizer= optimizer, loss=loss_fn, metrics=metrics)
```

```python
model.fit(x_train_scaled, y_train, 
          epochs=50, 
          batch_size=8, 
          validation_split=0.1
          )
```

```python
loss , accuracy = model.evaluate(x_test_scaled, y_test)
```
