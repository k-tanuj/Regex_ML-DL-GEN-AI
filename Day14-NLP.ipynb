```python
def remove_punc(text):
    for char in exclude:
        text = text.replace(char, '')
    return text
```

```python
chat_words = {'u2':'you too', 'brb':'be right back', 'idk':"I don't know", 'smh':'shaking my head'}
```

```python
def chat_convo(text):
    new_text = []
    for word in text.split():
        if word.lower() in chat_words:
            new_text.append(chat_words[word.lower()])
        else:
            new_text.append(word)
    return " ".join(new_text)
```

```python
chat_convo("idk what to do, brb")
```

```python
import nltk
nltk.download('stopwords')
from nltk.corpus import stopwords
```

```python
STOPWORDS = set(stopwords.words('english'))
```

```python
#manually work
def remove_stopwords(text):
    new_text = []
    for word in text.split():
        if word not in STOPWORDS:
            new_text.append(word)
    x = new_text[:]
    new_text.clear()
    return " ".join(x)

remove_stopwords("This is a sample sentence and with some stop words.")

```

```python
# Inflection : the process of changing the form of a word to express different grammatical features such as tense, case, voice, aspect, person

#Stemming : the process of reducing a word to its base or root form, which may not be a valid word in the language but serves as a common representation for related words. Stemming is often used in information retrieval and natural language processing tasks to group together different forms of a word.
```

```python
from nltk.stem.porter import PorterStemmer
```

```python
ps = PorterStemmer()

def stem_words(text):
    return " ".join([ps.stem(word) for word in text.split()])

```

```python
sample = "running runner runs ran"

stem_words(sample)
```

```python
sample2 = "walk walks walked walking"
stem_words(sample2)
```

```python
# the issue of stemming is that it may produce non-words or words that are not valid in the language. For example, the stem of "running" is "run", which is a valid word, but the stem of "runner" is "runner", which is not a valid word. This can lead to confusion and errors in natural language processing tasks.
```

```python
# Lemmatization : the process of reducing a word to its base or root form, which is a valid word in the language. 
# Lemmatization takes into account the context and meaning of the word, and it uses a dictionary or vocabulary to find the correct base form. 
# Lemmatization is often used in natural language processing tasks to improve the accuracy of text analysis and information retrieval.
```

```python
from nltk.stem import WordNetLemmatizer
import nltk
nltk.download('wordnet')
```

```python
wordnet_lemmatizer = WordNetLemmatizer()
```

```python
sentence = "The striped bats are hanging on their feet for best runner. why are they doing this? I am not sure. I will have to ask them."
punctuation = string.punctuation

sentence_words = nltk.word_tokenize(sentence)

for word in sentence_words:
    if word in sentence_words:
        sentence_words.remove(word)


print("{0:20}{1:20}".format("Word", "Lemma"))

for word in sentence_words:
    print("{0:20}{1:20}".format(word, wordnet_lemmatizer.lemmatize(word)))
```
