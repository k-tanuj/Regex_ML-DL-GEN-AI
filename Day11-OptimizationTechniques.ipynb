# Gradient Descent for Linear Regression

## Equation of a Straight Line

The equation of a straight line is:

y = mx + b

where:

- m = weight (slope)
- b = bias/intercept

---

## Loss Function (Mean Squared Error)

The Mean Squared Error (MSE) is:

L = (1/n) * Σ(y_actual - y_pred)^2

For a single training example (n = 1):

L = (y_actual - y_pred)^2

Substituting y_pred = mx + b:

L = (y - (mx + b))^2

L = (y - mx - b)^2

---

## Calculating Gradient with Respect to Bias (b)

Given:

L = (y - mx - b)^2

Applying the chain rule:

1. dL/db = 2(y - mx - b)[dy/db - d(mx)/db - db/db]

2. dL/db = 2(y - mx - b)[0 - 0 - 1]

3. dL/db = -2(y - mx - b)

Therefore:

dL/db = -2(y - mx - b)

---

## Calculating Gradient with Respect to Weight (m)

Given:

L = (y - mx - b)^2

Applying the chain rule:

1. dL/dm = 2(y - mx - b)[dy/dm - d(mx)/dm - db/dm]

2. dL/dm = 2(y - mx - b)[0 - x - 0]

3. dL/dm = -2x(y - mx - b)

Therefore:

dL/dm = -2x(y - mx - b)

---

## Updating the Parameters

Let:

lr = 0.01

### Update Bias

b_new = b_old - lr * (dL/db)

### Update Weight

m_new = m_old - lr * (dL/dm)

---

## Final Equation of the Line

y = m_new * x + b_new

---

## Summary

Loss Function:
L = (y - mx - b)^2

Gradient w.r.t Bias:
dL/db = -2(y - mx - b)

Gradient w.r.t Weight:
dL/dm = -2x(y - mx - b)

Parameter Updates:
b_new = b_old - lr * (dL/db)

m_new = m_old - lr * (dL/dm)

Final Equation:
y = m_new * x + b_new
