# Introduction To Deep Learning

##### Ml models works on less complex data, but DL models works on complex datasets.
##### DL -> Artificial Neural Network System 

## Workflow of DL:

    --> Data ingestion -> data cleaning and preprocessing -> model sequential architecture
    
        ->Model Sequential architecture:
            1. input layer(neurons, activation fns, input_dimensions)
            2. hidden layer(neurons, activation fns)
            3. output layer(neurons, activation fns based on target)

        -> model.compile(loss,optimizer,metrics)
        -> model training
        -> model prediction
        -> model performance evaluation

##### *Activation functions* are mathematical functions that is used for finding non-lineraity from the data.
    -> when data is numerical we will use :Relu,tanh
    -> when data is categorical we will use :sigmoid(2 sub-categories),softmax(for more than 2 sub-categories)

#### *Optimizers*  are used for minimizing the loss function(actual - predicted)
    1.SGD + Momentum -> stucks on local minima while oscillating around it
    2.SGD+Momentum+NAG -> overcomes the problem of local minima and oscillation by looking ahead of the current position
    3.AdaGrad -> provides the solution of automatic learning rate but cannot find proper solution.
    4.RMSProp -> Resolves the issue of automatic learning rate by adding an constant epsilon
    5.Adam (Momentum + RMSprop)-> Combines the advantages of both Momentum and RMSProp by using moving averages of the gradients and squared gradients to adapt the learning rate for each parameter.

```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv("Day12 data\\Attrition.csv")
```

```python
df.head()
```

```python
import tensorflow as tf
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.metrics import accuracy_score
```

```python
x= df.drop(columns=["Attrition"],axis=1)
y = df["Attrition"]
```

```python
x.info()
```

```python
categorical_features  = ['BusinessTravel', 'Department', 'EducationField','Gender','JobRole','MaritalStatus','Over18','OverTime']
```

```python
x_encoded = pd.get_dummies(x, columns = categorical_features, 
                           drop_first=True)
```

```python
x_encoded.head()
```

```python
x_encoded = x_encoded.astype(int)
```

```python
lb = LabelEncoder()
```

```python
y_encoded = lb.fit_transform(y)
y_encoded
```

```python
x_train, x_test, y_train, y_test = train_test_split(x_encoded, y_encoded, test_size=0.2, random_state=42)
```

```python
sc = StandardScaler()
```

```python
x_train_scaled = sc.fit_transform(x_train)
x_test_scaled = sc.transform(x_test)
```

```python
model = tf.keras.Sequential([
    tf.keras.layers.Dense(units = 128 ,activation='relu' ,input_dim=x_train_scaled.shape[1]), #input layer
    tf.keras.layers.Dropout(0.3),
    tf.keras.layers.Dense(units=64 ,activation='relu'), #hidden layer
    tf.keras.layers.Dropout(0.3),
    tf.keras.layers.Dense(units = 1 ,activation='sigmoid') #output layer
    ])
```

```python
#Complie the model
```

```python
loss_fn = tf.keras.losses.BinaryCrossentropy()
metrics = ['accuracy']
```

```python
learning_rate = 0.001
momentum = 0.9
optimizer = tf.keras.optimizers.SGD(
    learning_rate=learning_rate, 
    momentum=momentum,
    nesterov=True
    )
```

```python
model.compile(optimizer= optimizer, loss=loss_fn, metrics=metrics)
```

```python
#Model training
```

```python
model.fit(x_train_scaled, y_train, 
          epochs=50, 
          batch_size=8, 
          validation_split=0.1
          )
```

```python
loss , accuracy = model.evaluate(x_test_scaled, y_test)
```
