```python
# Term Freqquency

## Inverse Document Frequency (IDF)
```

```python
# TF(t,d) = (no. of occurunces of term t in document d)/ (total no. of terms in document d)

# example :
#d1 = 'people watch cricket'
#tf(people,d1) = 1/3

#d2 = 'player was player'
#tf(player,d2) = 2/3
```

```python
#IDF(t) = log_e(total no. of documents / no. of documents with term t in it)

#example :

#d1 = 'people watch cricket'
#d2 = 'cricket was cricket'
#d3 = 'people write cricket'
#d4 = 'cricket play comment'

#tf(people) = 4/2

#tf(cricket) = 4/4
```

# NER (NAME ENTITY RECOGNITION)
    NER  is a NLP technique used to find important entities in a text. It is used to extract information from unstructured text data. It can be used to extract names of people, organizations, locations, dates, and other important information from text data.
    Why do we use NER? -> to convert unstructured text data into structured data. 
    Main Reasons: Automatically extract imp information, reduce manual work, help machines understand cpntext, improve search and retrieval, enhance data analysis, enable automation, support decision making, facilitate knowledge discovery, improve user experience, enable personalization, support compliance and regulation.

```python
import spacy
nlp = spacy.load('en_core_web_sm')
```

```python
text = "Barack Obama was born in Hawaii and became the 44th president of the United States."

doc = nlp(text)

for ent in doc.ents:
    print(ent.text,"->",ent.label_)
```

```python
text2 = "Apple is looking at buying U.K. startup for $1 billion"

doc2 = nlp(text2)

for ent in doc2.ents:
    print(ent.text,"->",ent.label_)
```

```python
#create a pattern  that will mask or replace mobile no. , account no. , email id  from customer complaint, use regular expression.
```

```python
import re

def mask (text):
    
    text = re.sub(r'\b\d{10}\b','<MOBILE_NO>',text)
    text = re.sub(r'\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Z|a-z]{2,}\b','<EMAIL_ID>',text)
    text = re.sub(r'\b\d{9,18}\b','<ACCOUNT_NO>',text)
    return text


```

```python
text  = """
Hello, my name is Rohan. My mobile number is 9876543210.
My bank account number is 123456789012345.
Please contact me at rohan123@gmail.com regarding my issue.
"""
```

```python
mask(text)
```
